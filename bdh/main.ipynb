```python
def extract_numbers(res):
    res_t = int(res[0], 2)
    res_n = int(res[1:5], 2)
    res_car = int(res[5:10], 2)
    res_b = int(res[10:14], 2)
    res_a = int(res[14:18], 2)
    res_x_mult = int(res[18:22], 2)
    res_x = int(res[22:], 2)

    print(f"t = {res_t}, n = {res_n}, car = {res_car}, b = {res_b}, a = {res_a}, x_mult = {res_x_mult}, x = {res_x}")
```

```python
def parse_output(output):
    # in this string find "{'000000000000000000000100011': 1024}" and extract numbers

    res = output.split("\n")
    for i in range(len(res)):
        if len(res[i]) > 29 and res[i][0] == "{":
            extract_numbers(res[i][2:29])
        else:
            print(res[i])
```
